```python
import pandas as pd
import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

from statsmodels.tsa.holtwinters import (
    SimpleExpSmoothing,
    Holt,
    ExponentialSmoothing
)

from statsmodels.tsa.arima.model import ARIMA

from statsmodels.tsa.statespace.sarimax import SARIMAX

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("final_data_for_vehicle_forecasting.csv")

df['Month'] = pd.to_datetime(df['Month'])

df.head()
```

```python
TRAIN_END = pd.Timestamp("2024-12-01")

TEST_START = pd.Timestamp("2025-01-01")

TEST_END = pd.Timestamp("2025-03-01")
```

```python
#Evaluation Function
def evaluate_forecast(actual, forecast):

    mae = mean_absolute_error(actual, forecast)

    rmse = np.sqrt(
        mean_squared_error(actual, forecast)
    )

    mape = (
        mean_absolute_percentage_error(
            actual,
            forecast
        ) * 100
    )

    accuracy = 100 - mape

    return mae, rmse, mape, accuracy
```

```python
results = []
```

```python
#Main Forecasting Loop
for halb in df['Halb'].unique():

    df_h = df[df['Halb'] == halb]

    for engine in df_h['engine_type'].unique():

        df_e = df_h[
            df_h['engine_type'] == engine
        ]

        for model in df_e['map_model'].unique():

            df_m = df_e[
                df_e['map_model'] == model
            ]

            for vehicle in df_m['Vehicle_Type'].unique():

                df_v = df_m[
                    df_m['Vehicle_Type'] == vehicle
                ]
```

```python
#Monthly Series
agg = (
    df_v.groupby('Month')['total_demand']
        .sum()
        .sort_index()
)
```

```python
#Train test split
train = agg[
    agg.index <= TRAIN_END
]

test = agg[
    (agg.index >= TEST_START)
    &
    (agg.index <= TEST_END)
]
```

```python
print(len(train))
print(len(test))
```

```python
#ML Models
try:

    forecast = [
        train.tail(3).mean()
    ] * 3

    mae, rmse, mape, acc = (
        evaluate_forecast(
            test,
            forecast
        )
    )

    results.append([
        halb,
        engine,
        model,
        vehicle,
        "Moving Average",
        mae,
        rmse,
        mape,
        acc
    ])

except:
    pass
```

```python
try:

    model_ses = (
        SimpleExpSmoothing(train)
        .fit()
    )

    forecast = model_ses.forecast(3)

    mae, rmse, mape, acc = (
        evaluate_forecast(
            test,
            forecast
        )
    )

    results.append([
        halb,
        engine,
        model,
        vehicle,
        "SES",
        mae,
        rmse,
        mape,
        acc
    ])

except:
    pass
```

```python
try:

    model_holt = Holt(train).fit()

    forecast = model_holt.forecast(3)

    mae, rmse, mape, acc = (
        evaluate_forecast(
            test,
            forecast
        )
    )

    results.append([
        halb,
        engine,
        model,
        vehicle,
        "Holt",
        mae,
        rmse,
        mape,
        acc
    ])

except:
    pass
```

```python
try:

    model_hw = (
        ExponentialSmoothing(
            train,
            trend='add',
            seasonal='add',
            seasonal_periods=12
        )
        .fit()
    )

    forecast = model_hw.forecast(3)

    mae, rmse, mape, acc = (
        evaluate_forecast(
            test,
            forecast
        )
    )

    results.append([
        halb,
        engine,
        model,
        vehicle,
        "Holt Winters",
        mae,
        rmse,
        mape,
        acc
    ])

except:
    pass
```

```python
try:

    model_arima = (
        ARIMA(
            train,
            order=(1,1,1)
        )
        .fit()
    )

    forecast = model_arima.forecast(3)

    mae, rmse, mape, acc = (
        evaluate_forecast(
            test,
            forecast
        )
    )

    results.append([
        halb,
        engine,
        model,
        vehicle,
        "ARIMA",
        mae,
        rmse,
        mape,
        acc
    ])

except:
    pass
```

```python
try:

    model_sarima = (
        SARIMAX(
            train,
            order=(1,1,1),
            seasonal_order=(1,1,1,12)
        )
        .fit(disp=False)
    )

    forecast = model_sarima.forecast(3)

    mae, rmse, mape, acc = (
        evaluate_forecast(
            test,
            forecast
        )
    )

    results.append([
        halb,
        engine,
        model,
        vehicle,
        "SARIMA",
        mae,
        rmse,
        mape,
        acc
    ])

except:
    pass
```

```python
results_df = pd.DataFrame(
    results,
    columns=[
        'Halb',
        'Engine',
        'Model',
        'Vehicle_Type',
        'Model_Name',
        'MAE',
        'RMSE',
        'MAPE',
        'Accuracy'
    ]
)
```

```python
best_models = (
    results_df
    .sort_values('MAPE')
    .groupby(
        [
            'Halb',
            'Engine',
            'Model',
            'Vehicle_Type'
        ]
    )
    .first()
    .reset_index()
)
```

```python
results_df.to_csv(
    "all_model_results.csv",
    index=False
)

best_models.to_csv(
    "best_models.csv",
    index=False
)
```
