```python
overall_series = (
    df.groupby('Month')['total_demand']
      .sum()
      .sort_index()
)

print(overall_series)
```

```python
train = overall_series.loc[:'2024-12-01']

test = overall_series.loc['2025-01-01':'2025-03-01']
```

```python
print(train.shape)
print(test.shape)

print(test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

import numpy as np

def evaluate(actual, forecast):

    mae = mean_absolute_error(
        actual,
        forecast
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            forecast
        )
    )

    mape = (
        mean_absolute_percentage_error(
            actual,
            forecast
        ) * 100
    )

    accuracy = 100 - mape

    return mae, rmse, mape, accuracy
```

```python
moving_avg = train.tail(3).mean()

forecast = [moving_avg] * 3
```

```python
print("Actual")
print(test.values)

print()

print("Forecast")
print(forecast)
```

```python
mae, rmse, mape, acc = evaluate(
    test,
    forecast
)

print("MAE:", mae)
print("RMSE:", rmse)
print("MAPE:", mape)
print("Accuracy:", acc)
```

```python
results = []
```

```python
moving_avg = train.tail(3).mean()

forecast_ma = [moving_avg] * 3

mae, rmse, mape, acc = evaluate(
    test,
    forecast_ma
)

results.append([
    'Moving Average',
    mae,
    rmse,
    mape,
    acc
])
```

```python
from statsmodels.tsa.holtwinters import SimpleExpSmoothing

ses_model = SimpleExpSmoothing(
    train
).fit()

forecast_ses = ses_model.forecast(3)

mae, rmse, mape, acc = evaluate(
    test,
    forecast_ses
)

results.append([
    'SES',
    mae,
    rmse,
    mape,
    acc
])
```

```python
from statsmodels.tsa.holtwinters import Holt

holt_model = Holt(
    train
).fit()

forecast_holt = holt_model.forecast(3)

mae, rmse, mape, acc = evaluate(
    test,
    forecast_holt
)

results.append([
    'Holt',
    mae,
    rmse,
    mape,
    acc
])
```

```python
from statsmodels.tsa.arima.model import ARIMA

arima_model = ARIMA(
    train,
    order=(1,1,1)
).fit()

forecast_arima = arima_model.forecast(
    steps=3
)

mae, rmse, mape, acc = evaluate(
    test,
    forecast_arima
)

results.append([
    'ARIMA',
    mae,
    rmse,
    mape,
    acc
])
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

sarima_model = SARIMAX(
    train,
    order=(1,1,1),
    seasonal_order=(1,1,1,12)
).fit(disp=False)

forecast_sarima = sarima_model.forecast(
    steps=3
)

mae, rmse, mape, acc = evaluate(
    test,
    forecast_sarima
)

results.append([
    'SARIMA',
    mae,
    rmse,
    mape,
    acc
])
```

```python
results_df = pd.DataFrame(
    results,
    columns=[
        'Model',
        'MAE',
        'RMSE',
        'MAPE',
        'Accuracy'
    ]
)

results_df.sort_values(
    'MAPE'
)
```

```python
ml_df = overall_series.reset_index()

ml_df.columns = ['Month', 'Demand']

ml_df.head()
```

```python
ml_df['Year'] = ml_df['Month'].dt.year

ml_df['Month_Num'] = ml_df['Month'].dt.month

ml_df['Quarter'] = ml_df['Month'].dt.quarter
```

```python
ml_df['lag_1'] = ml_df['Demand'].shift(1)

ml_df['lag_2'] = ml_df['Demand'].shift(2)

ml_df['lag_3'] = ml_df['Demand'].shift(3)
```

```python
ml_df['rolling_mean_3'] = (
    ml_df['Demand']
    .rolling(3)
    .mean()
)
```

```python
ml_df = ml_df.dropna()

ml_df.head()
```

```python
print(ml_df.shape)

ml_df.head()
```

```python
train_ml = ml_df[
    ml_df['Month'] <= '2024-12-01'
]

test_ml = ml_df[
    ml_df['Month'] >= '2025-01-01'
]
```

```python
features = [
    'Year',
    'Month_Num',
    'Quarter',
    'lag_1',
    'lag_2',
    'lag_3',
    'rolling_mean_3'
]

X_train = train_ml[features]

y_train = train_ml['Demand']

X_test = test_ml[features]

y_test = test_ml['Demand']
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(
    X_train,
    y_train
)

forecast_lr = lr_model.predict(
    X_test
)

mae, rmse, mape, acc = evaluate(
    y_test,
    forecast_lr
)

results.append([
    'Linear Regression',
    mae,
    rmse,
    mape,
    acc
])
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(
    X_train,
    y_train
)

forecast_rf = rf_model.predict(
    X_test
)

mae, rmse, mape, acc = evaluate(
    y_test,
    forecast_rf
)

results.append([
    'Random Forest',
    mae,
    rmse,
    mape,
    acc
])
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

xgb_model.fit(
    X_train,
    y_train
)

forecast_xgb = xgb_model.predict(
    X_test
)

mae, rmse, mape, acc = evaluate(
    y_test,
    forecast_xgb
)

results.append([
    'XGBoost',
    mae,
    rmse,
    mape,
    acc
])
```

```python
results_df = pd.DataFrame(
    results,
    columns=[
        'Model',
        'MAE',
        'RMSE',
        'MAPE',
        'Accuracy'
    ]
)

results_df = results_df.sort_values(
    'MAPE'
)

results_df
```
